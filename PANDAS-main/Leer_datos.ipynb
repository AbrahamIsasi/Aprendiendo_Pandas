## <font color=#33A5FF>****2. Funciones básicas de lectura de ficheros****<tfont>

**index_col** = Cambiar el índice

```python
# Opción index_col
pd.read_excel('titanic.xlsx', index_col=0)
```

**nrows** = utilizar las n primeras filas

```python
# Opción nrows (para traer las filas  quen quiero)
pd.read_excel('titanic.xlsx', nrows=10)
```

**usecols**: Utilizar columnas específicas

```python
# Opción usecols ( para solo traer las columnas que quiere mostrar)
pd.read_excel('titanic.xlsx', usecols=['Survived',	'Clase',	'Nombre',	'Sexo',	'Edad'])
```

**skiprows**: saltarse filas de abajo hacia arriba

```python
# Opción skiprows
#sirve para salltarte filas para ver
```

**sheet_name**: Seleccionar las pestañas del libro excel

```python
# Opción sheet_name (para leer un hoja en especifico de excel)
pd.read_excel('titanic.xlsx', sheet_name=0)
```

**header**: Se utiliza cuando las columnas no tienen encabezado

```python
# Opción header ( para COLCAR ENCABEZAMOS A UNA TABLA)
pd.read_excel('titanic.xlsx', header=None , names= ['Id_Pasajero',	'Survived'	,'Clase'	,'Nombre'	,'Sexo',	'Edad'	,'Familiares'	,'Tiquete',	'Tarifa',	'Cabina',	'Embarcado'], sheet_name="Pasajeros_1")
```

## <font color=#33A5FF>****3. Funciones básicas de almacenamiento****<tfont>

```python
# Ruta de la carpeta donde quiero guardar mis archivos exportados
titanic = pd.read_excel('titanic.xlsx', header=None, names=[
        'Id_Pasajero','Survived','Clase','Nombre','Sexo','Edad',
        'Familiares','Tiquete','Tarifa','Cabina','Embarcado'
    ],
    sheet_name="Pasajeros_1"
)
# Exportar archivo formato excel 
titanic.to_excel('titanic_limpio.xlsx', index=False)
```

```python
# Exportar archivo formato excel 
titanic.to_excel('titanic_limpio.xlsx', index=False)
```

```python
# Exportar archivo formato excel sin índices
```

```python
# Exportar archivo formato .csv 
titanic.to_csv('titanic_nuevo.csv')
```

```python
# Exportar archivo formato .txt 
titanic.to_csv('titanic_nuevo.txt')
```

```python
# Exportar archivo formato .html 
titanic.to_html('titanic_nuevo.html')
```
